```python
# 代码实现 02_手动实现逻辑门.py
import torch

def or_gate(x1, x2):
    w1 = 1.0
    w2 = 1.0
    z = x1 * w1 + x2 * w2
    output = 1 if z >= 1 else 0
    return output

print("--- 与门 (OR Gate) 运行结果 ---")
print(f"输入 (0, 0) -> 预测输出: {or_gate(0, 0)}")
print(f"输入 (0, 1) -> 预测输出: {or_gate(0, 1)}")
print(f"输入 (1, 0) -> 预测输出: {or_gate(1, 0)}")
print(f"输入 (1, 1) -> 预测输出: {or_gate(1, 1)}")
```
